### Soru 1: Kargo Ücreti

Kullanıcıdan paketin ağırlığını (kg), gönderim mesafesini (km) ve hızlı kargo isteyip istemediğini (`e/h`) al. Ücreti şu tarifeye göre hesapla:

| Ağırlık | Ücret |
| :--- | :--- |
| 2 kg'a kadar (2 dahil) | 50 TL |
| 2-5 kg (5 dahil) | 80 TL |
| 5-10 kg (10 dahil) | 120 TL |
| 10 kg üstü | 120 TL + 10 kg'ı aşan her kg için 10 TL |

- Mesafe 500 km'den fazlaysa ücrete **%20** ekle.
- Hızlı kargo seçildiyse ücreti **1.5** ile çarp.
- Ağırlık ya da mesafe 0 veya negatifse ücret hesaplama, `'Geçersiz ağırlık ya da mesafe.'` yaz.

**Örnek Çıktı:**
```text
Paketin ağırlığı (kg): 12.5
Mesafe (km): 600
Hızlı kargo olsun mu? (e/h): e
Kargo ücreti: 261.00 TL
```

> **Kontrol:** Kodunu sınır değerlerle de dene: 2 kg 50 TL, 2.1 kg 80 TL tutmalı.

````python
agirlik = float(input("Paketin ağırlığı (kg): "))
mesafe = float(input("Mesafe (km): "))
hizlikargo = input("Hızlı kargo ister misiniz? (e/h): ")

if agirlik <= 0 or mesafe <= 0:
    print("Geçersiz ağırlık veya mesafe.")
else:
    if agirlik <= 2:
        ucret = 50
    elif agirlik <= 5:
        ucret = 80
    elif agirlik <= 10:
        ucret = 120
    else:
        ucret = 120 + (agirlik - 10) * 10

    if mesafe > 500:
        ucret = ucret * 1.20

    if hizlikargo == "e":
        ucret == ucret * 1.5

    print(f"Kargo ücreti: {ucret:.2f} TL")
    

    
````

### Soru 2: Döngüler

İki kısım, tek hücrede:

1. Kullanıcıdan bir sayı al ve 1'den 10'a kadar çarpım tablosunu `for` ve `range()` ile yazdır.
2. Kullanıcı `'q'` yazana kadar `while` ile sayı iste. Sayı olmayan bir girişte `'Geçersiz giriş'` yazıp devam et (`continue`), `'q'` gelince döngüden çık (`break`). Sonunda girilen sayıların **toplamını** ve **ortalamasını** yazdır.

**Örnek Çıktı (2. kısım):**
```text
Sayı gir (bitirmek için q): 3
Sayı gir (bitirmek için q): 5
Sayı gir (bitirmek için q): x
Geçersiz giriş, tekrar dene.
Sayı gir (bitirmek için q): 7

Toplam: 15 | Ortalama: 5.00
```

````python
"""
1.Kısım
"""
sayi = int(input("Çarpım tablosu için bir sayı giriniz:"))

for i in range(1,11):
    print(sayi, "x", i, "=", sayi * i)

"""
2.Kısım
"""
toplam = 0
adet = 0
while True:
    giris = input("Sayı gir (bitirmek için q): ")
    if giris == "q":
        break

    if not giris.replace(".", "",1).isdigit():
        print("Geçersiz giriş, lütfen tekrar deneyin. ")
        continue

    sayi = float(giris)
    toplam = toplam + sayi
    adet = adet + 1

    if adet > 0:
        ortalama = toplam / adet
        print(f"Toplam: {toplam:g} | Ortalama: { ortalama :.2f}")
    else:
        print("Geçerli bir sayı hiç girilmedi.")
````

### Soru 3: List Comprehension, zip ve enumerate

1. 1 ile 30 arasındaki çift sayıların karelerinden oluşan listeyi tek satırlık bir `list comprehension` ile oluştur ve yazdır.
2. `sicakliklar = [12, 15, 9, 18, 21, 17, 14]` listesindeki Celsius değerlerini list comprehension ile Fahrenheit'a çevir ($F = C \times 9/5 + 32$) ve bir basamağa yuvarla.
3. `isimler = ['Ali', 'Ayşe', 'Can']` ve `yaslar = [20, 19, 21]` listelerini `zip()` ile eşleştir, `enumerate()` ile numaralandırarak yazdır.

**Örnek Çıktı (2. ve 3. kısım):**
```text
[53.6, 59.0, 48.2, 64.4, 69.8, 62.6, 57.2]
1. Ali - 20 yaşında
2. Ayşe - 19 yaşında
3. Can - 21 yaşında
```

````python
"""
1-1 ile 30 arasındaki çift sayıların karelerinden oluşan liste.
"""
kareler = [sayi ** 2 for sayi in range(1, 31) if sayi % 2 == 0]
print(kareler)

"""
2-Sıcaklıklar listesindeki Celcius değerlerini Fahrenheit'a çevirme.
"""
sicakliklar = [12, 15, 9, 18, 21, 17, 14]

fahrenheit = [round(sicaklik * 9/5 + 32,1)
for sicaklik in sicakliklar]
print(fahrenheit)

"""
3-İsimler listesi üzerindeki değişiklikler.
"""
isimler = ["Ali", "Ayşe", "Can"]
yaslar = [20, 19, 21]
for sira, (isim, yas) in enumerate(zip(isimler, yaslar), start=1):
 print(f"{sira}. {isim} - {yas} yaşında")

````

### Soru 4: Fonksiyonlar

Aşağıdaki fonksiyonları yaz ve her birini en az bir kez çağırıp sonucunu yazdır. Fonksiyonlar sonucu ekrana yazmasın, `return` ile döndürsün.

1. `sesli_harf_say(metin)`: Metindeki sesli harf sayısını döndürür. `sesli_harf_say('Python öğreniyorum')` sonucu `6` olmalı.
2. `en_uzun_kelime(cumle)`: Cümledeki en uzun kelimeyi döndürür; eşitlikte ilk geleni seçer. `'Kod yazmak pratik ister'` için sonuç `'yazmak'`.
3. `istatistik(*sayilar)`: İstediğin kadar sayı alır; en küçük, en büyük ve ortalamayı birlikte döndürür. `istatistik(4, 8, 15, 16, 23, 42)` sonucu `(4, 42, 18.0)`.
4. `selamla(isim, dil='tr')`: Dil 'tr' ise 'Merhaba, Ayşe!', 'en' ise 'Hello, Ayşe!' döndürür. Dil verilmezse Türkçe selamlar.
5. Fonksiyon yazmadan: `['kod', 'python', 'döngü', 'if', 'fonksiyon', 'liste']` listesinden `filter()` ve `lambda` ile 4 harften uzun kelimeleri seç, `map()` ve `lambda` ile büyük harfe çevir.

````python
"""
1-Sesli harf say.
"""

def sesliharfsay(metin):
    sesliharfler = "aeıioöuüAEIİOÖUÜ"
    adet = 0

    for harf in metin:
        if harf in sesliharfler:
            adet = adet + 1

    return adet

print(sesliharfsay("Python öğreniyorum"))

"""
2-En uzun kelime(cümle).
"""
def enuzunkelime(cumle):
    kelimeler = cumle.split()
    enuzun = kelimeler[0]

    for kelime in kelimeler:
        if len(kelime) > len(enuzun):
            enuzun = kelime

    return enuzun

print(enuzunkelime("Kod yazmak pratik ister"))


"""
3-İstatistik(*sayılar).
"""
def istatistik(*sayilar):
    enkucuk = min(sayilar)
    enbuyuk = max(sayilar)
    ortalama = sum(sayilar) / len(sayilar)

    return enkucuk, enbuyuk, ortalama

print(istatistik(4, 8, 15, 16, 23, 42))


""" 
4-Selamla.
"""
def selamla(isim, dil="tr"):
    if dil == "tr":
        return f"Merhaba, {isim}!"
    elif dil == "en":
        return f"Hello, {isim}!"
    else:
        return f"Desteklenmeyen dil: {dil}"


print(selamla("Ayşe"))
print(selamla("Ayşe", "en"))



"""
5-Fonksiyon yazmadan tamamla.
"""

kelimeler = ["kod", "python","döngü", "if", "fonksiyon", "liste"]
uzunkelimeler = filter(lambda kelime: len(kelime) > 4, kelimeler)
buyukkelimeler = map(lambda kelime: kelime.upper(), uzunkelimeler)

print(list(buyukkelimeler))
````

### Soru 5: Adam Asmaca

Bilgisayarın tuttuğu kelimeyi harf harf tahmin ettiğin bir oyun yaz.

- Program en az 10 kelimelik bir listeden rastgele bir kelime seçsin: `kelime = random.choice(kelimeler)`. `random` modülünü 3. haftada göreceksin; şimdilik bu satırı olduğu gibi kullanabilirsin.
- Kelime başta `_ _ _ _` olarak gösterilsin; doğru tahmin edilen harfler yerlerine yazılsın.
- Oyuncunun 6 yanlış tahmin hakkı olsun. Her turda kalan hak ve denenen harfler gösterilsin.
- Daha önce denenen bir harf tekrar girilirse uyarı verilsin ve hak düşmesin.
- Tek harf olmayan girişler (boş, birden fazla karakter, rakam) geçersiz sayılsın.
- Kelime tamamlanınca kazandınız, haklar bitince kaybettiniz mesajı çıksın ve kelime gösterilsin.
- Kod en az iki fonksiyondan oluşsun.

**Örnek Çıktı:**
```text
Adam Asmaca'ya hoş geldin!

_ _ _ _ _
Kalan hak: 6 | Denenen harfler:
Bir harf tahmin et: a
Yanlış!

_ _ _ _ _
Kalan hak: 5 | Denenen harfler: a
Bir harf tahmin et: ö
Doğru!

_ ö _ _ _
Kalan hak: 5 | Denenen harfler: a ö
Bir harf tahmin et: ö
'ö' harfini zaten denedin.
```

````python
import random

def kelime_goster(kelime, tahminler):
    sonuc = ""

    for harf in kelime:
        if harf in tahminler:
            sonuc += harf
        else:
            sonuc += "_"

        sonuc += " "

    return sonuc



def adam_asmaca():
    kelimeler = ["python", "bilgisayar", "yazılım", "kodlama", "üniversite",
                 "ekran", "klavye", "mühendislik", "endüstri", "bilişim","İlke",
                 "Yıldız", "programlama", "matematik", "fizik", "lineer cebir", 
                 "genel sanat tarihi", "astronomi ve uzay bilimleri"]
    kelime = random.choice(kelimeler)


    tahminler = []
    hak = 6


    print("Adam Asmaca Oyununa Hoş Geldiniz!")


    while hak > 0:
        print("\nKelime:", kelime_goster(kelime, tahminler))
        print("Kalan Hakkınız:", hak)


        harf = input("Bir harf giriniz:").lower()


        if len(harf) !=1:
            print("Lütfen yalnızca bir tane harf giriniz!")
            continue

        if harf in tahminler:
            print("Bu harfi zaten önceden tahmin ettiniz!")
            continue

        tahminler.append(harf)

        if harf in kelime:
            print("Doğru tahmin!")
        else:
            print("Yanlış tahmnin!")
            hak -= 1


        tamamlandi = True


        for harf_kelime in kelime:
            if harf_kelime not in tahminler:
                tamamlandi = False



        if tamamlandi:
            print("\nTebrikler! Kelimeyi buldunuz:", kelime)
            break


    if hak == 0:
        print("\nKaybettiniz!")
        print("Gizli kelime:", kelime)


adam_asmaca()




# Yaparken hem bilgisyarım hem de beynim yanıyordu az kalsın :)



````
